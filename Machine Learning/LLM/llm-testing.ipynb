```python
# Parse HDBSCAN_clusters_after_training into a structured format
# ── Parse HDBSCAN cluster context into KB entries ─────────────
import re

def parse_hdbscan_clusters(raw: str) -> list[dict]:
    """
    Parse the HDBSCAN_clusters_after_training string into
    individual KB entries, one per cluster.
    """
    entries = []
    # Split on "Cluster X |" boundaries
    blocks = re.split(r'(?=Cluster -?\d+ \| n=)', raw.strip())

    for block in blocks:
        block = block.strip()
        if not block:
            continue

        # Extract cluster ID and size
        header = re.match(r'Cluster (-?\d+) \| n=(\d+)', block)
        if not header:
            continue

        cluster_id = int(header.group(1))
        n          = int(header.group(2))

        # Extract top paths block
        paths_match = re.search(r'Top paths:\n(.*?)(?=Avg request_count)', block, re.DOTALL)
        top_paths   = paths_match.group(1).strip() if paths_match else ""

        # Extract metrics
        req_count = re.search(r'Avg request_count: ([\d.]+)', block)
        ratio_4xx = re.search(r'Avg ratio_4xx: ([\d.]+)', block)

        avg_req   = float(req_count.group(1)) if req_count else 0.0
        avg_4xx   = float(ratio_4xx.group(1)) if ratio_4xx else 0.0

        # Pull top 5 paths as a clean list
        path_lines = [l.strip() for l in top_paths.splitlines() if l.strip()]
        top_5      = path_lines[:5]

        entries.append({
            "id":    f"hdbscan_cluster_{cluster_id}",
            "title": f"HDBSCAN Cluster {cluster_id} — training fingerprint (n={n})",
            "content": (
                f"Cluster {cluster_id} contains {n} IP windows from training data. "
                f"Top observed paths: {'; '.join(top_5)}. "
                f"Average request count per window: {avg_req:.2f}. "
                f"Average 4xx ratio: {avg_4xx:.4f}. "
                f"Full path list: {top_paths.replace(chr(10), ' | ')}."
            )
        })

    return entries

CLUSTER_CONTEXT_ENTRIES = parse_hdbscan_clusters(HDBSCAN_clusters_after_training)

print(f"Parsed {len(CLUSTER_CONTEXT_ENTRIES)} cluster context entries")
for e in CLUSTER_CONTEXT_ENTRIES[:3]:
    print(f"  {e['id']}: {e['title']}")
    print(f"  {e['content'][:120]}...")
```

```python
import requests
import json
from collections import defaultdict

CF_PHASE     = "http_request_firewall_custom"
CF_PHASE_URL = f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets/phases/{CF_PHASE}/entrypoint"
CF_HEADERS   = {
    "Authorization": f"Bearer {CLOUDFLARE_API_TOKEN}",
    "Content-Type":  "application/json"
}

MAX_EXPRESSION_LENGTH = 4096
RULE_BUDGET           = 5  # Cloudflare Free tier hard limit

# ── Helpers ──────────────────────────────────────────────────

def _get_ruleset() -> tuple[str | None, list[dict]]:
    """
    Returns (ruleset_id, existing_rules).
    Creates the entry point ruleset if it doesn't exist.
    """
    try:
        r = requests.get(CF_PHASE_URL, headers=CF_HEADERS)
        if r.status_code == 200:
            result = r.json()["result"]
            return result["id"], result.get("rules", [])
        elif r.status_code == 404:
            print("⚠ Entry point ruleset not found — creating...")
            r2 = requests.post(
                f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets",
                headers=CF_HEADERS,
                json={"kind": "zone", "name": "Custom rules", "phase": CF_PHASE, "rules": []}
            )
            r2.raise_for_status()
            return r2.json()["result"]["id"], []
        else:
            r.raise_for_status()
    except requests.exceptions.RequestException as e:
        print(f"✗ Could not get/create ruleset: {e}")
        return None, []

def _validate_expression(expression: str) -> tuple[bool, str]:
    if not expression.strip():
        return False, "Expression is empty."
    if len(expression) > MAX_EXPRESSION_LENGTH:
        return False, f"Expression exceeds {MAX_EXPRESSION_LENGTH} chars ({len(expression)})."
    cf_fields = [
        "http.request.uri", "http.request.uri.path", "http.request.uri.query",
        "http.host", "ip.src", "ip.geoip.country",
        "http.user_agent", "http.request.method", "cf.threat_score"
    ]
    if not any(f in expression for f in cf_fields):
        return False, "Expression doesn't reference any known Cloudflare field."
    return True, "ok"

def _find_existing_rule(action: str, rules: list[dict]) -> dict | None:
    """Find an existing [AUTO] rule for this action tier so we can update it."""
    for rule in rules:
        desc = rule.get("description", "").lower()
        if "[auto]" in desc and rule.get("action") == action:
            return rule
    return None

def _build_merged_expression(patterns: list[str]) -> str:
    """
    OR-chain a list of bare path patterns into a single Cloudflare expression.
    Each pattern becomes: (http.request.uri.path contains "pattern")
    Falls back to raw expression if the pattern already contains CF field names.
    """
    parts = []
    for p in patterns:
        if "http." in p or "ip." in p or "cf." in p:
            # Already a full CF expression — wrap in parens and use as-is
            parts.append(f"({p})" if not p.startswith("(") else p)
        else:
            parts.append(f'(http.request.uri.path contains "{p}")')
    return " or ".join(parts)

# ── Core tool function ────────────────────────────────────────

def create_consolidated_firewall_rules(cluster_groups: list[dict]) -> dict:
    """
    Create or update Cloudflare WAF rules, consolidated by action tier.
    Covers all detected clusters using the minimum number of rules.

    Parameters
    ----------
    cluster_groups : list of dicts, each with:
        {
          "action":       str,         # block | managed_challenge | js_challenge | challenge
          "cluster_ids":  list[int],   # cluster IDs covered by this group
          "patterns":     list[str],   # path patterns OR full CF expressions to OR-chain
          "description":  str,         # human label e.g. "[AUTO] CRITICAL block rule"
          "priority":     int | None   # 1=highest, omit to append
        }

    Returns
    -------
    dict: summary of actions taken per group
    """
    valid_actions = {"block", "challenge", "js_challenge", "managed_challenge"}
    ruleset_id, existing_rules = _get_ruleset()
    if not ruleset_id:
        return {"error": "Could not resolve ruleset ID"}

    used_rules   = len([r for r in existing_rules if "[auto]" not in r.get("description","").lower()])
    auto_rules   = len([r for r in existing_rules if "[auto]" in r.get("description","").lower()])
    budget_left  = RULE_BUDGET - used_rules
    results      = {}

    print(f"\n── Rule budget: {used_rules} manual + {auto_rules} [AUTO] = {used_rules+auto_rules}/{RULE_BUDGET} used")
    print(f"── Groups to process: {len(cluster_groups)}\n")

    for group in cluster_groups:
        action      = group.get("action", "")
        cluster_ids = group.get("cluster_ids", [])
        patterns    = group.get("patterns", [])
        description = group.get("description", f"[AUTO] {action.upper()} consolidated rule")
        priority    = group.get("priority")

        # Ensure [AUTO] prefix
        if "[auto]" not in description.lower():
            description = f"[AUTO] {description}"

        # Validate action
        if action not in valid_actions:
            results[action] = {"error": f"Invalid action '{action}'"}
            continue

        # Build merged expression
        if not patterns:
            results[action] = {"error": "No patterns provided", "clusters": cluster_ids}
            continue

        merged = _build_merged_expression(patterns)

        # Validate
        valid, reason = _validate_expression(merged)
        if not valid:
            results[action] = {"error": reason, "clusters": cluster_ids}
            continue

        # Check if an [AUTO] rule for this action already exists → UPDATE it
        existing = _find_existing_rule(action, existing_rules)

        if existing:
            rule_id         = existing["id"]
            existing_expr   = existing.get("expression", "")

            # Append only new patterns that aren't already covered
            new_parts = [
                p for p in patterns
                if p not in existing_expr
            ]

            if not new_parts:
                print(f"  ✓ [{action.upper()}] rule {rule_id} already covers all patterns — skipping")
                results[action] = {
                    "status":   "unchanged",
                    "rule_id":  rule_id,
                    "clusters": cluster_ids
                }
                continue

            # Build updated expression by appending new patterns
            additional  = _build_merged_expression(new_parts)
            updated_expr = f"{existing_expr} or {additional}"

            valid, reason = _validate_expression(updated_expr)
            if not valid:
                results[action] = {"error": f"Updated expression invalid: {reason}", "clusters": cluster_ids}
                continue

            rules_url = f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets/{ruleset_id}/rules/{rule_id}"
            try:
                r = requests.patch(
                    rules_url,
                    headers=CF_HEADERS,
                    json={"expression": updated_expr, "description": description, "enabled": True}
                )
                r.raise_for_status()
                print(f"  ✓ [{action.upper()}] rule {rule_id} UPDATED — +{len(new_parts)} new patterns "
                      f"({len(updated_expr)} chars / {MAX_EXPRESSION_LENGTH})")
                results[action] = {
                    "status":        "updated",
                    "rule_id":       rule_id,
                    "patterns_added": len(new_parts),
                    "expression_length": len(updated_expr),
                    "clusters":      cluster_ids
                }
            except requests.exceptions.RequestException as e:
                print(f"  ✗ [{action.upper()}] update failed: {e}")
                results[action] = {"error": str(e), "clusters": cluster_ids}

        else:
            # No existing rule for this action — CREATE new one
            if budget_left <= 0:
                msg = f"Rule budget exhausted ({RULE_BUDGET}/{RULE_BUDGET}). Cannot create new rule for {action}."
                print(f"  ✗ {msg}")
                results[action] = {"error": msg, "clusters": cluster_ids}
                continue

            rule_body = {
                "expression":  merged,
                "action":      action,
                "description": description,
                "enabled":     True,
            }
            if priority is not None:
                rule_body["position"] = {"index": priority}

            rules_url = f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets/{ruleset_id}/rules"
            try:
                r = requests.post(rules_url, headers=CF_HEADERS, json=rule_body)
                r.raise_for_status()
                rules    = r.json().get("result", {}).get("rules", [])
                new_rule = next((ru for ru in reversed(rules) if ru.get("expression") == merged), None)
                rule_id  = new_rule.get("id") if new_rule else "unknown"
                budget_left -= 1
                print(f"  ✓ [{action.upper()}] rule {rule_id} CREATED — {len(patterns)} patterns "
                      f"({len(merged)} chars / {MAX_EXPRESSION_LENGTH}) | budget left: {budget_left}")
                results[action] = {
                    "status":            "created",
                    "rule_id":           rule_id,
                    "patterns_covered":  len(patterns),
                    "expression_length": len(merged),
                    "clusters":          cluster_ids
                }
            except requests.exceptions.RequestException as e:
                print(f"  ✗ [{action.upper()}] create failed: {e}")
                results[action] = {"error": str(e), "clusters": cluster_ids}

    return results
```

```python
def query_knowledgebase(query: str, top_k: int = 3) -> str:
    """Semantic search over the D3FEND/MITRE KB — called by the model as a tool."""
    # q_vec = np.array(client.feature_extraction(query, model=EMBED_MODEL))
    q_vec = np.array(vo.embed([query], model="voyage-3-lite", input_type="document").embeddings)[0]
    q_vec = q_vec / np.linalg.norm(q_vec)

    scores  = kb_vectors @ q_vec
    top_idx = np.argsort(scores)[::-1][:top_k]

    results = []
    for i in top_idx:
        if i < len(kb_entries):
            e = kb_entries[i]
            results.append(
                f"[KB-ID: {e['id']}] {e['title']}\n"
                f"Content: {e['content']}"
            )
    return "\n\n---\n\n".join(results)
```

```python
def delete_firewall_rule(rule_id: str) -> bool:
    """Delete a rule from the http_request_firewall_custom phase by rule ID."""
    ruleset_id, _ = _get_ruleset()
    if not ruleset_id:
        print("✗ Could not get ruleset ID")
        return False

    url = f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets/{ruleset_id}/rules/{rule_id}"
    try:
        r = requests.delete(url, headers=CF_HEADERS)
        r.raise_for_status()
        print(f"✓ Deleted rule: {rule_id}")
        return True
    except requests.exceptions.RequestException as e:
        print(f"✗ Failed to delete {rule_id}: {e}")
        return False

def delete_all_auto_rules() -> None:
    """Delete only [AUTO] rules created by the pipeline, leaving manual rules intact."""
    _, rules = _get_ruleset()
    auto_rules = [r for r in rules if "[auto]" in r.get("description", "").lower()]
    if not auto_rules:
        print("No [AUTO] rules found.")
        return
    for rule in auto_rules:
        delete_firewall_rule(rule["id"])

def check_token_permissions() -> None:
    """Check what permissions the current API token has."""
    url = f"https://api.cloudflare.com/client/v4/accounts/{CLOUDFLARE_ACCOUNT_ID}/tokens/verify"
    try:
        r = requests.get(url, headers=CF_HEADERS)
        r.raise_for_status()
        print(json.dumps(r.json(), indent=2))
    except requests.exceptions.RequestException as e:
        print(f"✗ {e}")

def get_ruleset_metadata() -> None:
    """Check who owns/created the ruleset — managed rulesets can't be edited via API."""
    ruleset_id, rules = _get_ruleset()
    url = f"https://api.cloudflare.com/client/v4/zones/{CLOUDFLARE_ZONE_ID}/rulesets/{ruleset_id}"
    try:
        r = requests.get(url, headers=CF_HEADERS)
        result = r.json().get("result", {})
        print(f"Ruleset ID   : {result.get('id')}")
        print(f"Kind         : {result.get('kind')}")      # 'zone' = yours, 'managed' = Cloudflare's
        print(f"Phase        : {result.get('phase')}")
        print(f"Description  : {result.get('description')}")
        print(f"Last updated : {result.get('last_updated')}")
        print(f"Total rules  : {len(result.get('rules', []))}")
    except requests.exceptions.RequestException as e:
        print(f"✗ {e}")

# Verify
check_token_permissions()
get_ruleset_metadata()
```
